### Feature engineering (rolling and shifting mean)

```python
combined_df = combined_df.sort_values(['Store', 'Date'])
combined_df
```

```python
combined_df['sales_lag_1'] = combined_df.groupby('Store')["Sales"].shift(1)
combined_df
```

```python
combined_df['sales_roll_mean_2'] = combined_df.groupby('Store')["Sales"].shift(1).rolling(window=2).mean()
```

```python
combined_df['sales_roll_mean_7'] = combined_df.groupby('Store')["Sales"].shift(1).rolling(window=7).mean()
```

### Summary — Baseline & First Model

**Baseline (per-store, per-day-of-week average):** MAE ≈ 1425 (evaluated on open-only rows)

**Random Forest (100 trees):** MAE ≈ 838 on validation — a ~43% reduction in average error
over the baseline, evaluated on the same open-only rows for a fair comparison.

**Feature importance:** `sales_roll_mean_7` (~56%) and `sales_lag_1` (~16%) dominate,
confirming that recent sales history is the strongest signal for near-term demand.
`Promo` (~13%) is the next most important, consistent with promotions being a well-known
sales driver in retail. Calendar features (`DayOfWeek`, `Month`) and holiday flags
contribute comparatively little — likely because the lag/rolling features already
implicitly capture weekly seasonality patterns.

**Limitations:** Predictions range ~803–28,540 vs. actual sales ranging ~569–41,551 —
the model likely underpredicts on unusually high-demand days, a common property of
tree-based ensembles that average toward typical patterns rather than extrapolating
to extremes. Holiday-specific accuracy hasn't been checked in isolation and would be
worth investigating further if extending this project.

### Predicting for multiple days

```python
store_id = 1
start_date = pd.Timestamp('2015-07-01')
horizon_days = 7
```

```python
store_history_df = train[(train['Store'] == store_id) & (train['Date'] < start_date)]
store_history_df = pd.concat([
    store_history_df,
    val[(val['Store'] == store_id) & (val['Date'] < start_date)]
])
store_history_df = store_history_df.sort_values('Date')
```

```python
horizon_dates = pd.date_range(start=start_date, periods=horizon_days)
actual_rows = test[(test['Store'] == store_id) & (test['Date'].isin(horizon_dates))].sort_values('Date')

promo_schedule = actual_rows['Promo'].tolist()
open_schedule = actual_rows["Open"].tolist()
```

```python
forecast_df = predict_horizon(
    store_history_df=store_history_df,
    model=model,
    encoder=encoder,
    feature_cols=feature_cols,
    start_date=start_date,
    horizon_days=horizon_days,
    promo_schedule=promo_schedule,
    open_schedule = open_schedule
)
```

```python
comparison = forecast_df.merge(
    actual_rows[['Date', 'Sales']], on='Date', how='left'
)
comparison['abs_error'] = (comparison['Predicted_Sales'] - comparison['Sales']).abs()
print(comparison)
```
